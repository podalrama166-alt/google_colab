**Q3**. Create a list of fruits. Add a new fruit using append() and display the updated list.

```python
list3 = ["Apple", "Banana", "Coconut", "Dragon Fruit", "Elderberry"]
print("The old list is : \n", list3)
list3.append("Orange")
print("The new list is : \n", list3)
```

**Q4**. Create a list of subjects. Insert "AI" at the third position and display the updated list.

**Q6**. Create a list:

[10, 20, 30, 40, 50]

Display:

- Length of the list
- Maximum value
- Minimum value
- Sum of all elements

```python
list4 = [10, 20, 30, 40, 50]
print("The length of the list is :", len(list4))
print("The maximum value of the list is :", max(list4))
print("The minimum value of the list is :", min(list4))
print("The sum of all elements of the list is :", sum(list4))
```

**Q7**. Take 5 numbers from the user and store them in a list. Display the complete list.

```python
list6 = []
print("Here we enter 5 numbers ")
for i in range(5):
    number = input(f"Enter {i}th number :")
    list6.append(number)
print(list6)
```
