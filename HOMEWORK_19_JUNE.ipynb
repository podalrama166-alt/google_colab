**Q19**. Create two sets:

A = {1, 2, 3, 4}

B = {3, 4, 5, 6}

Display:

- Union
- Intersection

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print("The set A is :\n", A)
print("The set B is :\n", B)
print("The union of set A and set B is :\n", A.union(B))
print("The intersection of set A and set B is :\n", A.intersection(B))
```

**Q30**. Create a dictionary where:

- Key = Student Name
- Value = Marks

Display all students who scored more than 80 marks.

```python
Student_dict = {
    "Rama" : 79,
    "Shyam" : 90,
    "Krushna" : 67,
    "Prabin" : 94,
    "Bikash" : 82
}
print("The student dictionary is : \n", Student_dict)
print("The names of the students who scored more than 80 marks are : ")
for name in Student_dict:
    if Student_dict[name] > 80:
        print(name)
```
