# What is an Instance Method in Python?

An **instance method** is a method that **belongs to an object (instance) of a class**. It can access and modify the object's data (instance variables).

An instance method always has **`self`** as its **first parameter**.

- `self` refers to the **current object** that calls the method.

**EXAMPLES OF INSTANCE METHOD**

Example 1: Display Student Information

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student("Rama", 22)
s1.display()
```

Example 2: Bank Deposit

```python
class Bank:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("New Balance:", self.balance)

acc = Bank(1000)
acc.deposit(500)
```

Example 3: Bank Withdrawal

```python
class Bank:

    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        self.balance -= amount
        print("Remaining Balance:", self.balance)

acc = Bank(2000)
acc.withdraw(700)
```

Example 4: Car Speed

```python
class Car:

    def __init__(self):
        self.speed = 0

    def accelerate(self):
        self.speed += 20
        print("Speed:", self.speed)

car = Car()

car.accelerate()
car.accelerate()
```

Example 5: Employee Salary

```python
class Employee:

    def __init__(self, salary):
        self.salary = salary

    def increase_salary(self):
        self.salary += 5000
        print("Updated Salary:", self.salary)

emp = Employee(30000)
emp.increase_salary()
```

Example 6: Rectangle Area

```python
class Rectangle:

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        print("Area:", self.length * self.width)

r = Rectangle(10, 5)
r.area()
```

EXAMPLES OF CLASS METHOD

Example 1: Display School Name

```python
class Student:

    school = "Pace Computer Academy"

    @classmethod
    def display_school(cls):
        print("School Name:", cls.school)

Student.display_school()
```

Example 2: Change School Name

```python
class Student:

    school = "ABC School"

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

Student.change_school("Pace Computer Academy")

print(Student.school)
```

Example 3: Employee Company Name

```python
class Employee:

    company = "Google"

    @classmethod
    def display_company(cls):
        print("Company:", cls.company)

Employee.display_company()
```

Example 4: Change Company Name

```python
class Employee:

    company = "Google"

    @classmethod
    def change_company(cls, company_name):
        cls.company = company_name

Employee.change_company("Microsoft")

print(Employee.company)
```

Example 5: Count Total Students

```python
class Student:

    total_students = 0

    def __init__(self):
        Student.total_students += 1

    @classmethod
    def display_total_students(cls):
        print("Total Students:", cls.total_students)

s1 = Student()
s2 = Student()
s3 = Student()

Student.display_total_students()
```

Example 6: Change Tax Rate

```python
class Product:

    tax = 18

    @classmethod
    def change_tax(cls, new_tax):
        cls.tax = new_tax

Product.change_tax(12)

print(Product.tax)
```

EXAMPLES OF STATIC METHOD

Example 1: Welcome Message

```python
class Student:

    @staticmethod
    def welcome():
        print("Welcome to Python!")

Student.welcome()
```

Example 2: Add Two Numbers

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
```

Example 3: Find Maximum Number

```python
class Math:

    @staticmethod
    def maximum(a, b):
        if a > b:
            return a
        return b

print(Math.maximum(15, 8))
```

Example 4: Check Even or Odd

```python
class Number:

    @staticmethod
    def is_even(num):
        if num % 2 == 0:
            print("Even Number")
        else:
            print("Odd Number")

Number.is_even(18)
Number.is_even(7)
```

Example 5: Celsius to Fahrenheit

```python
class Temperature:

    @staticmethod
    def celsius_to_fahrenheit(c):
        return (c * 9/5) + 32

print(Temperature.celsius_to_fahrenheit(25))
```

Example 6: Calculate Circle Area

```python
class Circle:

    @staticmethod
    def area(radius):
        return 3.14 * radius * radius

print(Circle.area(7))
```
